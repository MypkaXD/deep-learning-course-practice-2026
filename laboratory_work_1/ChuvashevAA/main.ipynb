```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class MLP:
    def __init__(self, in_dim, hidden_dim, out_dim, bias = True):

        self.bias = bias

        self.w1 = np.empty(shape=(in_dim, hidden_dim), dtype=np.float32)
        self.w2 = np.empty(shape=(hidden_dim, out_dim), dtype=np.float32)
        
        if self.bias:
            self.b1 = np.empty(shape=(hidden_dim, 1), dtype=np.float32)
            self.b2 = np.empty(shape=(out_dim, 1), dtype=np.float32)
        
    def forward(self, x):

        x1 = x.T * self.w1
        if self.bias:
            x1 += self.b1
        
        x2 = x1.T * self.w2
        if self.bias:
            x2 += self.b2

        return x2

    
```
